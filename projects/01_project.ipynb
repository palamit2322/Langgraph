```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict,Literal
from dotenv import load_dotenv
import random
load_dotenv()
```

```python
class State(TypedDict):
    name:str
    message:str
```

```python
graph=StateGraph(State)
```

```python
def name_node(state:State):
    return {
        "message": f"my name is {state['name']}"
    }
```

```python
def cricket_node(state:State)->State:
    return {
        "message":f"my name is {state['name']} and I play cricket"
    }

def football_node(state:State)->State:
    return {
         "message":f"my name is {state['name']} and I play football"
    }
```

```python
def selection_node(state:State)->Literal["cricket","football"]:
    if random.random()>0.5:
        return "cricket"
    else:
        return "football"
```

```python
graph.add_node("name",name_node)
graph.add_node("cricket",cricket_node)
graph.add_node("football",football_node)
```

```python
graph.add_edge(START,"name")
graph.add_conditional_edges("name",selection_node,{
    "cricket":"cricket",
    "football":"football"
})
graph.add_edge("cricket",END)
graph.add_edge("football",END)
```

```python
graph.compile()
```

```python
workflow=graph.compile()
```

```python
print(workflow.get_graph().draw_mermaid())
```

```python
response=workflow.invoke({"name":"Amit Pal"})
print(type(response))
print(response["message"])
```
